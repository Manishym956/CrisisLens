```python
import pandas as pd

buzz_fake = pd.read_csv(
    "archive6/BuzzFeed_fake_news_content.csv"
)

buzz_real = pd.read_csv(
    "archive6/BuzzFeed_real_news_content.csv"
)

politifact_fake = pd.read_csv(
    "archive6/PolitiFact_fake_news_content.csv"
)

politifact_real = pd.read_csv(
    "archive6/PolitiFact_real_news_content.csv"
)
```

```python
buzz_fake["label"] = 1
buzz_real["label"] = 0

politifact_fake["label"] = 1
politifact_real["label"] = 0
```

```python
df = pd.concat([
    buzz_fake,
    buzz_real,
    politifact_fake,
    politifact_real
])
```

```python
print(df.columns)
```

```python
df["content"] = (
    df["title"].fillna("")
    + " "
    + df["text"].fillna("")
)
```

```python
df = df[["content", "label"]]

df.dropna(inplace=True)

df = df[df["content"].str.len() > 20]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df["content"],
    df["label"],
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    stop_words="english",
    max_features=5000
)

X_train_tfidf = vectorizer.fit_transform(
    X_train
)

X_test_tfidf = vectorizer.transform(
    X_test
)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression()

model.fit(
    X_train_tfidf,
    y_train
)
```

```python
from sklearn.metrics import accuracy_score

preds = model.predict(X_test_tfidf)

print(
    accuracy_score(y_test, preds)
)
```

```python
ISOT_Fake=pd.read_csv("News_Dataset/Fake.csv")
ISOT_True=pd.read_csv("News_Dataset/True.csv")
```

```python
fake = ISOT_Fake.sample(2000)
true = ISOT_True.sample(2000)
```

```python
fake["label"] = 1
true["label"] = 0
```

```python
import pandas as pd

df = pd.concat([fake, true])
```

```python
df["content"] = (
    df["title"].fillna("") +
    " " +
    df["text"].fillna("")
)
```

```python
df = df[["content", "label"]]
```

```python
df = df.sample(frac=1).reset_index(drop=True)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    df["content"],
    df["label"],
    test_size=0.2,
    random_state=42
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    stop_words='english',
    max_features=5000,
    ngram_range=(1,2)
)

X_train_tfidf = vectorizer.fit_transform(X_train)

X_test_tfidf = vectorizer.transform(X_test)
```

```python
from sklearn.linear_model import PassiveAggressiveClassifier

model = PassiveAggressiveClassifier()

model.fit(
    X_train_tfidf,
    y_train
)
```

```python
from sklearn.metrics import accuracy_score

predictions = model.predict(X_test_tfidf)

accuracy = accuracy_score(
    y_test,
    predictions
)

print("Accuracy:", accuracy)
```

```python
import joblib

joblib.dump(
    model,
    "fake_news_model.pkl"
)

joblib.dump(
    vectorizer,
    "tfidf_vectorizer.pkl"
)

print("Model Saved Successfully")
```

```python
import pandas as pd
import joblib

from sklearn.feature_extraction.text import CountVectorizer
from sklearn.decomposition import LatentDirichletAllocation
```

```python
fake_df = pd.read_csv("News_Dataset/Fake.csv")
```

```python
fake_df = fake_df.sample(2000)
```

```python
fake_df["content"] = (
    fake_df["title"].fillna("")
    + " "
    + fake_df["text"].fillna("")
)
```

```python
fake_df = fake_df[["content"]]

fake_df.dropna(inplace=True)

fake_df = fake_df[
    fake_df["content"].str.len() > 50
]
```

```python
lda_vectorizer = CountVectorizer(
    stop_words='english',
    max_features=5000
)

X = lda_vectorizer.fit_transform(
    fake_df["content"]
)
```

```python
lda_model = LatentDirichletAllocation(
    n_components=5,
    random_state=42
)

lda_model.fit(X)
```

```python
words = lda_vectorizer.get_feature_names_out()

for index, topic in enumerate(lda_model.components_):

    print(f"\nTOPIC {index}")

    top_words = [
        words[i]
        for i in topic.argsort()[-10:]
    ]

    print(top_words)
```

```python
import pandas as pd
import numpy as np
import joblib

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
fake_df = pd.read_csv(
    "News_Dataset/Fake.csv"
)
```

```python
fake_df = fake_df.sample(2000)
```

```python
fake_df["content"] = (
    fake_df["title"].fillna("")
    + " "
    + fake_df["text"].fillna("")
)
```

```python
lda_model = joblib.load(
    "lda_model.pkl"
)

lda_vectorizer = joblib.load(
    "lda_vectorizer.pkl"
)
```

```python
X_lda = lda_vectorizer.transform(
    fake_df["content"]
)

topic_probs = lda_model.transform(X_lda)
```

```python
fake_df["trend_label"] = np.argmax(
    topic_probs,
    axis=1
)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    fake_df["content"],
    fake_df["trend_label"],
    test_size=0.2,
    random_state=42
)
```

```python
trend_vectorizer = TfidfVectorizer(
    stop_words='english',
    max_features=5000
)

X_train_tfidf = trend_vectorizer.fit_transform(
    X_train
)

X_test_tfidf = trend_vectorizer.transform(
    X_test
)
```

```python
trend_model = LogisticRegression(
    max_iter=1000
)

trend_model.fit(
    X_train_tfidf,
    y_train
)
```

```python
preds = trend_model.predict(
    X_test_tfidf
)

accuracy = accuracy_score(
    y_test,
    preds
)

print("Trend Accuracy:", accuracy)
```

```python
joblib.dump(
    trend_model,
    "trend_predictor.pkl"
)
```

```python
joblib.dump(
    trend_vectorizer,
    "trend_vectorizer.pkl"
)
```
